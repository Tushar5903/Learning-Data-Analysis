```python
import json
```

```python
#Load the data from the file
def load_data(filename):
    with open(filename,'r') as f:
        data = json.load(f)
    return data
```

```python
def find_friend_suggestion(user_id,data):
    user_friends = {}

    #find user friends
    for user in data['users']:
        user_friends[user['id']] = set(user['friends'])

    # Return nothing with no friends user
    if user_id not in user_friends:
        return []

    #find friends suggestion
    direct_friend = user_friends[user_id]
    suggestion = {}

    for friends in direct_friend:
        #find all mututal friend

        for mutual in user_friends[friends]:
            if mutual != user_id and mutual not in direct_friend:
                suggestion[mutual] = suggestion.get(mutual,0)+1

        sorted_suggestion = sorted(suggestion.items(), key=lambda x:x[1], reverse=True)
        return [user_id for user_id, _ in sorted_suggestion]

    
```

```python
data = load_data('massive_data.json')
user_id = 12
recommendation = find_friend_suggestion(user_id,data)
print(f"People you may also know {user_id}: {recommendation}")
```
