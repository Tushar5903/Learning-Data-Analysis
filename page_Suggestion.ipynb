```python
import json
```

```python
# Load the data from the file

def load_data(filename):
    with open(filename,'r') as f:
        data = json.load(f)
    return data
```

```python
# Function for suggesting the pages

def suggested_page(user_id,data):
    user_page = {}
    for user in data['users']:
        user_page[user['id']] = set(user['liked_pages'])

    #If the user didn't like any page
    if user_id not in user_page:
        return []

    user_liked_pages = user_page[user_id]
    page_suggestion = {}

    for other_user, pages in user_page.items():
        if other_user != user_id:
            shared_page = user_liked_pages.intersection(pages)
            for page in pages:
                if page not in user_liked_pages:
                    page_suggestion[page] = page_suggestion.get(page,0) + len(shared_page)

    sorted_pages= sorted(page_suggestion.items(), key=lambda x:x[1], reverse=True)
    return [page_id for page_id, score in sorted_pages if score >=1]
        
```

```python
data = load_data('massive_data.json')
user_id =11
suggestions = suggested_page(user_id,data)
print(f"Page suggetion for the {user_id}: {suggestions}");
```
