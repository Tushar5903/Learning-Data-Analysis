```python
import json
```

```python
#Load the data from the json file
def load_data(filename):
    with open(filename, 'r') as f:
        data = json.load(f)
        return data
```

```python
data = load_data('data.json')
```

```python
# write a function to display the user's friend and there likes
def display_user(data):
    id_to_name = {u['id']: u['name'] for u in data ['users']}
    id_to_page = {p['id']: p['name'] for p in data ['pages']}
    count = 1
    for user in data['users']:
        friend_name = [ id_to_name[fid] for fid in user['friends']]
        page_name = [id_to_page[pid] for pid in user['liked_pages']]
        print(f" ID {count}: {user['name']} is friends with {friend_name} and likes {page_name}" )
        count +=1
        
```

```python
infomation = display_user(data)
```
